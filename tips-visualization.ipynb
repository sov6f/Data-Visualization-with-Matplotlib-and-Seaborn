```python
# Create a histogram using Matplotlib
plt.figure(figsize=(8, 5))

plt.hist(df["total_bill"], bins=15, color="skyblue", edgecolor="black")

plt.title("Distribution of Total Bills", fontsize=16)
plt.xlabel("Total Bill ($)", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.savefig("visualizations/matplotlib_histogram.png", dpi=300)
plt.show()
```

```python
# Create a histogram using Seaborn
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="total_bill", bins=15, color="salmon", edgecolor="black")

plt.title("Distribution of Total Bills", fontsize=16)
plt.xlabel("Total Bill ($)", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.savefig("visualizations/seaborn_histogram.png", dpi=300)
plt.show()
```

```python
# Calculate average tip for each day
avg_tip = df.groupby("day", observed=True)["tip"].mean()

# Create a bar chart using Matplotlib
plt.figure(figsize=(8, 5))

plt.bar(avg_tip.index, avg_tip.values, color="skyblue", edgecolor="black")

plt.title("Average Tip by Day", fontsize=16)
plt.xlabel("Day", fontsize=12)
plt.ylabel("Average Tip ($)", fontsize=12)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.savefig("visualizations/matplotlib_bar_chart.png", dpi=300)
plt.show()
```

```python
# Create a bar chart using Seaborn
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="day", y="tip", color="salmon")

plt.title("Average Tip by Day", fontsize=16)
plt.xlabel("Day", fontsize=12)
plt.ylabel("Average Tip ($)", fontsize=12)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.savefig("visualizations/seaborn_bar_chart.png", dpi=300)
plt.show()
```

```python
# Create a box plot using Matplotlib
plt.figure(figsize=(8, 5))

days = df["day"].unique()
bill_data = [df[df["day"] == day]["total_bill"] for day in days]

plt.boxplot(bill_data, tick_labels=days)

plt.title("Total Bill Distribution by Day", fontsize=16)
plt.xlabel("Day", fontsize=12)
plt.ylabel("Total Bill ($)", fontsize=12)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.savefig("visualizations/matplotlib_box_plot.png", dpi=300)
plt.show()
```

```python
# Create a scatter plot using Matplotlib
plt.figure(figsize=(8, 5))

plt.scatter(df["total_bill"], df["tip"], color="skyblue", edgecolor="black", alpha=0.7)

plt.title("Total Bill vs Tip", fontsize=16)
plt.xlabel("Total Bill ($)", fontsize=12)
plt.ylabel("Tip ($)", fontsize=12)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.savefig("visualizations/matplotlib_scatter_plot.png", dpi=300)
plt.show()
```

```python
# Calculate the correlation matrix
corr = df.corr(numeric_only=True)

# Create a heatmap using Matplotlib
plt.figure(figsize=(8, 5))

plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, fontsize=10)
plt.yticks(range(len(corr.columns)), corr.columns, fontsize=10)

# Add correlation values to the cells
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")

plt.title("Correlation Heatmap", fontsize=16)

plt.tight_layout()
plt.savefig("visualizations/matplotlib_heatmap.png", dpi=300)
plt.show()
```

```python
# Create a heatmap using Seaborn
plt.figure(figsize=(8, 5))

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)

plt.title("Correlation Heatmap", fontsize=16)

plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.savefig("visualizations/seaborn_heatmap.png", dpi=300)
plt.show()
```
